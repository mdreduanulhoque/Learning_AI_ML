Creating SQL Database

```python
CREATE DATABASE SQL_ASSIGNMENT;

USE SQL_ASSIGNMENT;

CREATE TABLE employee(
    EmpID INT, 
    FirstName TEXT, 
    LastName TEXT, 
    Department TEXT, 
    Salary DOUBLE, 
    HireDate Timestamp
);
```

Q1. Write a query to display every employee and all their data.

```python
SELECT * FROM employee;
```

Q2. List only the FirstName, LastName, and Salary of every employee.

```python
SELECT FirstName, LastName, Salary FROM employee;
```

Q3. Show all employees who work in the 'IT' department.

```python
SELECT * FROM employee WHERE Department LIKE 'IT';
```

Q4. Retrieve employees with a salary greater than 6000.

```python
SELECT * FROM employee WHERE Salary > 6000;
```

Q5. List all employees ordered by HireDate from newest to oldest.

```python
SELECT * FROM employee ORDER BY HireDate ASC;
```

Q6. Show a list of all unique departments present in the table.

```python
SELECT DISTINCT Department FROM employee;
```

Q7. Find employees whose first name starts with ‘Aʼ.

```python
SELECT * FROM employee WHERE FirstName LIKE "A%";
```

Q8. Show employees whose salaries are between 4000 and 7000.

```python
SELECT * FROM employee WHERE Salary >= 4000 AND Salary <= 7000;
```

Q9. Find the average salary of all employees.

```python
SELECT AVG(Salary) AS avg_salary FROM employee;
```

Q10. List each department along with the number of employees, but only include departments with more than or euqals 3 employees.

```python
SELECT Department , COUNT(EmpID) FROM employee GROUP BY Department HAVING COUNT(EmpID) >= 3;
```
