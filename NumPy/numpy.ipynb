```python
# import
import numpy as np
import time
```

```python
# Create numpy array from python list
lst = [1, 2, 3, 4, 5, 6]
arr = np.array(lst)

print(arr)
print(type(arr))
```

```python
# Performance Compare
size = 100_000_000

py_lst = list(range(size))
p_start = time.time()
s_py_list = [x**2 for x in py_lst]
p_end = time.time()
print(f"Python list take {p_end-p_start} seconds")

np_array = np.array(py_lst)
np_start = time.time()
s_np_array = np_array**2
np_end = time.time()
print(f"Numpy array take {np_end-np_start} seconds")

print(f"python list is {(p_end-p_start)/(np_end-np_start)} times slower than numpy array")
```
