```python
# import
import numpy as np
import time
```

```python
# Create numpy array from python list
lst = [1, 2, 3, 4, 5, 6]
arr = np.array(lst)

print(arr)
print(type(arr))
```

```python
# Performance Compare
size = 100_000_000

py_lst = list(range(size))
p_start = time.time()
s_py_list = [x**2 for x in py_lst]
p_end = time.time()
print(f"Python list take {p_end-p_start} seconds")

np_array = np.array(py_lst)
np_start = time.time()
s_np_array = np_array**2
np_end = time.time()
print(f"Numpy array take {np_end-np_start} seconds")

print(f"python list is {(p_end-p_start)/(np_end-np_start)} times slower than numpy array")
```

```python
ar = np.array([1, 2, 3, 4, 5])
print(f"Type: {ar.dtype} || Shape: {ar.shape} || Content: {ar}")

ar1 = np.array([1, 2, 3, 4, 5, "hi"])
print(f"Type: {ar1.dtype} || Shape: {ar1.shape} || Content: {ar1}")

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(f"Type: {arr_2d.dtype} || Shape: {arr_2d.shape} || Content: {arr_2d}")
```

```python
# Build in functions
arr_z = np.zeros((2, 3), dtype="int64") 
print(f"Shape: {arr_z.shape} || Output: {arr_z}")

arr_o = np.ones((2, 3)) 
print(f"Shape: {arr_o.shape} || Output: {arr_o}")

arr_f = np.full((3, 4), 5)
print(f"Shape: {arr_f.shape} || Output: {arr_f}")

arr_a = np.arange(1, 20, 3)
print(f"Shape: {arr_a.shape} || Output: {arr_a}")

arr_l = np.linspace(0, 100, 5)
print(f"Shape: {arr_l.shape} || Output: {arr_l}")
```
