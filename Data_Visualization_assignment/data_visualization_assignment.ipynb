```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

Q1. For the following dataset generated using NumPy:

np.random.normal(loc=70, scale=10, size=200)

. Plot a histogram of scores         
. Draw a vertical line for the mean & try to interpret the results.

```python
df = np.random.normal(loc=70, scale=10, size=200)

plt.hist(df, bins = 10, color='skyblue', edgecolor="black", alpha = 0.8)
mean_score = np.mean(df)
plt.axvline(mean_score, color='red', linestyle='--', linewidth=0.6, label=f"Mean Score = {mean_score:.2f}", alpha=0.9)

plt.title("Histogram of scores")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

```python
# Interpretation of result
print(f"The data set contains {np.size(df)} values.\nThe mean of the score is {df.mean():.2f} which is close to the 'loc' value mentioned in the df creation.\nThe standard deviation of the dataset is {df.std():.2f} which is close to the scale mentioned in the begining")
```

Q2. For the Seaborn penguins dataset, do the following:

. Plot flipper_length_mm vs body_mass_g               
. Color by species

```python
ds = sns.load_dataset("penguins")

sns.scatterplot(data=ds, x = "flipper_length_mm", y= "body_mass_g", hue="species")
plt.title("flipper_length_mm vs body_mass_g")
plt.legend()
plt.show()
```
