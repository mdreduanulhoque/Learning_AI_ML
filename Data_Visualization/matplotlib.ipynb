```python
import matplotlib.pyplot as plt
import numpy as np
```

Line Plot

```python
# Line Plots

x = [1, 2, 3, 4, 5]
y = [6, 7, 8, 9, 10]

plt.plot(x, y)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("X vs Y graph")
plt.grid(True)
plt.show()
```

```python
# Styline the line

plt.plot(x, y, color="blue", linestyle="--", linewidth=2, marker = "o", markersize=6)
plt.show()
```

```python
# Multiple Line in single plot
y2 = [1, 9, 5, 3, 7]

plt.plot(x, y, label="Line 1")
plt.plot(x, y2, label="line 2")

plt.xlabel(("X Axis"))
plt.ylabel("Y Axis")
plt.legend()
plt.show()
```

```python
#Saving plot/ image
# plt.savefig("file_name")
```

```python
# Use a line plot when:

# Data points are ordered
# You want to show progression or trends
# You're comparing changes rather than individual categories
```

Bar Chart

```python
x = ['A', 'B', 'C', 'D', 'E']
y = [12, 18, 9, 15, 20]

plt.bar(x, y)
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Templete Chart")
plt.show()
```

```python
# Styling the bar chart

plt.bar(x, y, color="skyblue", width=0.6, edgecolor="black", linewidth=1.5)

plt.legend()
plt.show()
```

```python
# Horizontal Bar chart

plt.barh(x, y, color='skyblue', 
         height=0.6, edgecolor='black', 
         linewidth=1.5)

plt.xlabel("Values")
plt.ylabel("Categories")
plt.title("Horizontal Bar Chart")
plt.show()
```

```python
# Multiple bar in one chart

x = np.arange(5)
y1 = [10, 15, 7, 12, 15]
y2 = [8, 14, 9, 10, 7]

width = 0.35
plt.bar(x - width/2, y1, width, label="Group 1")
plt.bar(x + width/2, y2, width, label="Group 2")

plt.xticks(x, ['A', 'B', 'C', 'D', 'E'])
plt.xlabel("Categories")
plt.ylabel("Values")
plt.legend()
plt.show()
```

```python
# Use a bar chart when:

# . You are comparing different categories
# . The data is discrete, not continuous
# . You want clear visual comparison
```

Scatter Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 1]

plt.scatter(x, y)
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Simple Scatter Plot")
plt.grid(True)
plt.show()
```

```python
# Styling scatter plot
plt.scatter(x, y, s = 100, c= 'red', marker = 'o',
            alpha = 0.7)
plt.show()
```

```python
#Using a colormap

colors = np.array([10, 20, 30, 40, 50])

plt.scatter(x, y, c = colors, cmap='viridis')
plt.colorbar(label="Color Scale")
plt.show()
```

```python
# Multiple Scatter Plots

y2 = [5, 3, 4, 2, 1]

plt.scatter(x, y, label='Set 1')
plt.scatter(x, y2, label='Set 2')
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
# Use a scatter plot when:
# . You want to analyze the relationship between two variables
# . Data is numerical and continuous
# . You don't want to imply a trend by connecting points
```

Pie Chart

```python
sizes = [30, 20, 25, 25]
labels = ['A', 'B', 'C', 'D']

plt.pie(sizes, labels=labels)
plt.show()
```

```python
# Adding percentages

plt.pie(sizes,
    labels=labels,
    autopct='%1.1f%%') # shows percentage
plt.show()
```

```python
# Changing colors

colors = ['skyblue', 'lightgreen', 'pink', 'orange']

plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors)
plt.show()
```

```python
# Starting angle and shadow

plt.pie(sizes,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90, # rotate chart
    shadow=True) # adds shadow
plt.show()
```

```python
# styling with wedgeprops

plt.pie(sizes,
labels=labels,
autopct='%1.1f%%',
wedgeprops={'edgecolor': 'black', 'linewidth': 2})
```

```python
# Use a pie chart when:
# • You want to display part-to-whole relationships
# • You have categorical data
# • Data has few categories (4–6); too many slices make it hard to read
```

Histograms

```python
data = [1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 5, 5]

plt.hist(data, bins=5, color='skyblue', edgecolor='black')

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Simple Histogram")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.hist(data, bins=[1, 2, 3, 4, 5, 6], color='skyblue', edgecolor='black') # sequence
```

```python
# For horizontal histogram

plt.hist(data, orientation='horizontal')
```

```python
data2 = [2, 3, 3, 4, 4, 5, 5, 5, 6]
plt.hist(data, bins=5, alpha=0.5, label='Data 1', color='blue')
plt.hist(data2, bins=5, alpha=0.5, label='Data 2', color='red')
plt.legend()
```

```python
# Use a histogram when:
# • You want to show the distribution of continuous data.
# • You have to analyze frequency, spread, skewness, or patterns.
# • Working with large datasets to summarize trends.
```

Box Plots

```python
data = [7, 8, 5, 6, 9, 7, 8, 10, 4, 6]
plt.boxplot(data)
plt.ylabel("Values")
plt.title("Simple Box Plot")
plt.show()
```

```python
# horizontal Box

plt.boxplot(data, vert=False)
plt.xlabel("Values")
plt.show()
```

```python
# Multiple Box plot

data1 = [7, 8, 5, 6, 9, 7, 8, 10, 4, 6]
data2 = [5, 6, 7, 8, 5, 4, 6, 7, 5, 6]
plt.boxplot([data1, data2], labels=['Dataset 1', 'Dataset 2'])
plt.show()
```

```python
# Use box plot when:

# • Visualize spread and skewness of data
# • Identify outliers
# • Compare distributions across multiple groups
```

Stack Plots

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 2, 3, 4, 5]
y2 = [2, 1, 2, 1, 2]
plt.stackplot(x, y1, y2, labels=['Data1', 'Data2'], colors=['blue', 'green'])
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Basic Stack Plot")
plt.legend(loc='upper left')
plt.show()
```

```python
# Use a stack plot when you want to:
# • Show cumulative data over time
# • Compare multiple components
# • Highlight trends in parts and whole
```

Subplots

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 3, 5, 7, 11]
y2 = [1, 4, 2, 5, 3]
# 1 row, 2 columns, first subplot
plt.subplot(1, 2, 1) # (rows, columns, index)
plt.plot(x, y1, color='blue', marker='o')
plt.title("First Subplot")
plt.xlabel ("X")
plt.ylabel("Y1")
# 1 row, 2 columns, second subplot
plt.subplot(1, 2, 2)
plt.plot(x, y2, color='red', marker='s')
plt.title("Second Subplot")
plt.xlabel("X")
plt.ylabel("Y2")
plt.tight_layout()
plt.show()
```
