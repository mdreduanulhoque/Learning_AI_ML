```python
import matplotlib.pyplot as plt
```

```python
# Line Plots

x = [1, 2, 3, 4, 5]
y = [6, 7, 8, 9, 10]

plt.plot(x, y)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("X vs Y graph")
plt.grid(True)
plt.show()
```

```python
# Styline the line

plt.plot(x, y, color="blue", linestyle="--", linewidth=2, marker = "o", markersize=6)
plt.show()
```

```python
# Multiple Line in single plot
y2 = [1, 9, 5, 3, 7]

plt.plot(x, y, label="Line 1")
plt.plot(x, y2, label="line 2")

plt.xlabel(("X Axis"))
plt.ylabel("Y Axis")
plt.legend()
plt.show()
```
