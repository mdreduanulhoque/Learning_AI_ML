Series --> 1D labeled array
DataFrame --> 2D labled array

```python
import pandas as pd
import numpy as np
```

```python
# Pandas Series

#index based accessing
s =  pd.Series([23, 56, 89, 54])
print(s[3])
print(s[1])

#label based accessing
s = pd.Series([23, 56, 89, 54], index=["alex", "bob", "rahim", 'karim'])
print(s["rahim"])
print(s["alex"])
```

```python
# Pandas Series Properties

# 1. Homogenous
h = pd.Series([23, 56, 89, 54, "rafik"], index=["alex", "bob", "rahim", 'karim', 45])
print(h)  #all elements become same type

# 2. Vectorized Operations
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd. Series([10, 20, 30, 40, 50])
print(s1 + s2)

# 3. Handle Missing values with NaN

# 4. Mutable values, immutable size --> Values can be changed, if you try to change the size,
# it will create another copy and change the size, will not be reflected in main series
s3 = pd.Series([1, 2, 3, 4, 5])
s3[0] = 100
print(s3)
changed_s3 = s3.drop(0)
print(changed_s3)
print(s3)
```

```python
# Conversion dictionary to dataframe
info = {
    "Name" : ["Rakib", "Rahim", "Karim"],
    "Age" : [4, 7, 9]
}

df1 = pd.DataFrame(info)
print(df1)

# list of list to dataframe
df2 = pd.DataFrame([["Rakib", 4], ["Rahim", 7], ["Karim", 9]], columns=["Name", "Age"])
print(df2)

# np array to dataframe
np_array = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
df3 = pd.DataFrame(np_array, columns=["A", "B", "C", "D"])
print(df3)

## df.columns --> to get all columns of dataframe
## df.index --> to get all rows of dataframe

print(df2.index)
print(df2.columns)
```
