```python
import pandas as pd
import numpy as np
```

```python
# Pandas Series

#index based accessing
s =  pd.Series([23, 56, 89, 54])
print(s[3])
print(s[1])

#label based accessing
s = pd.Series([23, 56, 89, 54], index=["alex", "bob", "rahim", 'karim'])
print(s["rahim"])
print(s["alex"])
```

```python
# Pandas Series Properties

# 1. Homogenous
h = pd.Series([23, 56, 89, 54, "rafik"], index=["alex", "bob", "rahim", 'karim', 45])
print(h)  #all elements become same type

# 2. Vectorized Operations
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd. Series([10, 20, 30, 40, 50])
print(s1 + s2)

# 3. Handle Missing values with NaN

# 4. Mutable values, immutable size --> Values can be changed, if you try to change the size,
# it will create another copy and change the size, will not be reflected in main series
s3 = pd.Series([1, 2, 3, 4, 5])
s3[0] = 100
print(s3)
changed_s3 = s3.drop(0)
print(changed_s3)
print(s3)
```

```python
# Conversion dictionary to dataframe
info = {
    "Name" : ["Rakib", "Rahim", "Karim"],
    "Age" : [4, 7, 9]
}

df1 = pd.DataFrame(info)
print(df1)

# list of list to dataframe
df2 = pd.DataFrame([["Rakib", 4], ["Rahim", 7], ["Karim", 9]], columns=["Name", "Age"])
print(df2)

# np array to dataframe
np_array = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
df3 = pd.DataFrame(np_array, columns=["A", "B", "C", "D"])
print(df3)

## df.columns --> to get all columns of dataframe
## df.index --> to get all rows of dataframe

print(df2.index)
print(df2.columns)
```

## Global AIr Quality Dataset
https://www.kaggle.com/datasets/smeet888/global-air-quality-data15-days-hourly-50-cities

```python
df = pd.read_csv("globalAirQuality.csv")
df
```

```python
df.describe()
```

```python
# select value of a specific column/columns

print(df["country"])
print(df[["country", "city", "aqi"]])
```

```python
# access row/ row wise data
df.loc[1]  #loc --> label based
df.loc[[0, 3, 7]]
df.loc[0:4, ["city", "aqi"]]

df.iloc[0]  #iloc --> index based
df.iloc[[0, 3, 7]]
df.iloc[0:4, [2, 11]]

#same goes for at and iat
```

```python
"""
Logical operator in pandas

AND --- > &
OR ---> |
NOT ---> ~
"""
```

```python
#filtering data

df[df["aqi"]>100]
df[(df["aqi"]>100) & (df["temperature"]>30)]

#filtering data with chaining
df[df["aqi"]>100][["city", "aqi"]]
df[(df["aqi"]>100) & (df["temperature"]>30)][["city", "aqi"]]

#Accessing data after filtering
#After fileter, original label preserves in pandas, to retrieve with index, 
#use iloc[]
aqi_data = df[(df["aqi"]>100) & (df["temperature"]>30)][["city", "aqi"]]
aqi_data.iloc[0]
```

```python
# Filtering data using .query()
df.query("aqi > 157")
df.query("aqi > 157 & temperature < 20")

df.query("aqi > 157 & temperature < 20")[['city', 'aqi']]

aqi_filter = 36
temp_filter = 29
df.query("aqi > @aqi_filter & temperature < @temp_filter")[['city', 'aqi']]

#if you need a copy instead of a view, ask for it
dat = df.query("aqi > 157 & temperature < 20").copy()
dat
```

```python
# Handling Missing values
raw_data = pd.read_csv("raw_data.csv")
cleaned_data = raw_data.copy()

cleaned_data.isnull()
cleaned_data.isnull().sum()

# cleaned_data.dropna()  --> it will drop all the rows with atleast one missing values
# cleaned_data.dropna(axis = 1) --> it will drop all the columns with atleast one missing values

cleaned_data["age"] = cleaned_data["age"].fillna(cleaned_data["age"].mean())
cleaned_data["income"] = cleaned_data["income"].fillna(cleaned_data["income"].mean())

cleaned_data["country"] = cleaned_data["country"].bfill()  # it will fill null with the next data

cleaned_data["gender"] = cleaned_data["gender"].ffill()  # it will fill null with the previous data
```

```python
cleaned_data.isnull().sum()
```

```python
# Handle Duplicate Value
cleaned_data.duplicated()

cleaned_data["country"].duplicated()
cleaned_data[["country", "gender"]].duplicated()

cleaned_data.drop_duplicates() #it will create a view not any actual changes in main data
cleaned_data.drop_duplicates(inplace=True) #actually changes the data
cleaned_data
```

```python
# Dealing with data type
cleaned_data.columns
cleaned_data["age"].dtype
cleaned_data["age"] = cleaned_data["age"].astype("int64")
cleaned_data["age"].dtype
```
