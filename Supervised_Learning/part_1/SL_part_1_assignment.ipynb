```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import numpy as np
import matplotlib as plt
import seaborn as sns
```

```python
df = pd.read_csv("HousePricePrediction.csv")
df.head()
```

```python
print(df.isnull().sum())
df.dropna(inplace=True)
```

```python
print(df.isnull().sum())
```

```python
df["MSZoning"] = df["MSZoning"].map({"RL":0, "RM": 1, "C (all)": 2, "FV": 3, "RH": 4})
df["LotConfig"] = df["LotConfig"].map({"Inside":0, "FR2": 1, "Corner": 2, "culDSac": 3, "FR3": 4})
df["BldgType"] = df["BldgType"].map({"1Fam": 0, "2FmCon":1, "Duplex":2, "TwnhsE":3, "Twnhs":4})
```

```python
df["BldgType"].unique()
df.isnull().sum()
```

```python
df.drop(["Exterior1st"],axis=1, inplace=True)
df.drop(["LotConfig"],axis=1, inplace=True)
df.drop(["BldgType"],axis=1, inplace=True)
```

```python
X = df.drop(["SalePrice"], axis=1)
y = df["SalePrice"]

x_train, x_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.2)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)
y_pred
```

```python
r2_score_pred = r2_score(y_test, y_pred)
print("r2 score is :", r2_score_pred)
```
