```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv("insurance.csv")
df.head()
```

```python
df["sex"] = df["sex"].str.lower()
df["sex"] = df["sex"].map({"male": 1, "female" : 0})

df["smoker"] = df["smoker"].str.lower()
df["smoker"] = df["smoker"].map({"yes":1, "no": 0})
df
```

```python
#Interaction Features
df["age_smoker"] = df["age"] * df["smoker"]
df["bmi_smoker"] = df["bmi"] * df["smoker"]
df["age_bmi"] = df["age"] * df["bmi"]

X = df.drop(["charges"], axis=1)
y = df["charges"]

#one hot Encoding
X = pd.get_dummies(X, columns=["region"], drop_first=True, dtype=int)

x_train, x_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.2)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)
y_pred
```

```python
r2_score_pred = r2_score(y_test, y_pred)
print("r2 score is :", r2_score_pred)
```

```python
# Over fit vs under fit check

# training data accuracy low & test data accuracy low -> Underfitting
# training data accuracy very high & test data accuracy low -> Overfitting

y_train_pred = (model.predict(x_train))

r2_train = r2_score(y_train, y_train_pred)
r2_test = r2_score(y_test, y_pred)

print("Train data r2 value: ", r2_train)
print("Test data r2 value: ", r2_test)
```
