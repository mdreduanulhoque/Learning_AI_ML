```python
import pandas as pd
import numpy as np
```

```python
# Feature Enginnering - (Creating new column/ feature to help ML model to recognize pattern better)

df = pd.read_csv("raw_data.csv")
df2 = df.copy()

df2["tax"] = df["income"].apply(lambda x : "20%" if x > 60000 else "10%")
df2  # the function that takes another function as parameter is called higher order function

gender_map = {"Male":"M", "Female":"F", "Unknown":"U"}
df2["gender"] = df2["gender"].map(gender_map)
df2

df2 = df2.assign(new_income = df["income"] * 1.1)
df2

df2["country"] = df2["country"].replace("USA", "United States")
df2
```

```python
# data transformation

df2 = df2.rename(columns={"name":"Full Name"})
df2
df2 = df2.rename(columns={"Full Name":"name"})
df2

sorted_df2 = df2.sort_values("age")
sorted_df2
sorted_df2 = df2.sort_values("age", ascending=False)
sorted_df2
```
