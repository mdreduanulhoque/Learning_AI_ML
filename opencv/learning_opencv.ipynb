```python
import os
import cv2
import time
```

```python
# Read image
img_path = "bird.jpg"
img = cv2.imread(img_path)

print(img.shape)
```

```python
#write image
cv2.imwrite("output_bird.jpg", img)

# will be deleted after 5 second
time.sleep(5)
os.remove("output_bird.jpg")
```

```python
# Visualize image
cv2.imshow("image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Accessing web cam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    cv2.imshow("Webcam", frame)

    if cv2.waitKey(1) != -1:
        break

cap.release()
cv2.destroyAllWindows()
```

Image resizing

```python
img = cv2.imread("bird.jpg")

#pass height and width here
resized_img = cv2.resize(img, (339, 226))

# width, height, RGB
print(img.shape)
print(resized_img.shape)

cv2.imshow("img1", img)
cv2.imshow("img2", resized_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

Image Cropping

```python
img = cv2.imread("bird.jpg")
print(img.shape)

cropped_img = img[52:352, 100:578]
print(cropped_img.shape)

cv2.imshow("img1", img)
cv2.imshow("img_cropped", cropped_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
