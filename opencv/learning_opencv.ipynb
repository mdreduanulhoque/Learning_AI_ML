```python
import os
import cv2
import time
```

```python
# Read image
img_path = "bird.jpg"
img = cv2.imread(img_path)

print(img.shape)
```

```python
#write image
cv2.imwrite("output_bird.jpg", img)

# will be deleted after 5 second
time.sleep(5)
os.remove("output_bird.jpg")
```

```python
# Visualize image
cv2.imshow("image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Accessing web cam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    cv2.imshow("Webcam", frame)

    if cv2.waitKey(1) != -1:
        break

cap.release()
cv2.destroyAllWindows()
```

Image resizing

```python
img = cv2.imread("bird.jpg")

#pass height and width here
resized_img = cv2.resize(img, (339, 226))

# width, height, RGB
print(img.shape)
print(resized_img.shape)

cv2.imshow("img1", img)
cv2.imshow("img2", resized_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

Image Cropping

```python
img = cv2.imread("bird.jpg")
print(img.shape)

cropped_img = img[52:352, 100:578]
print(cropped_img.shape)

cv2.imshow("img1", img)
cv2.imshow("img_cropped", cropped_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

Changing Colorspace

```python
# Default colorspace: BGR
img = cv2.imread("bird.jpg")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

cv2.imshow("img", img)
cv2.imshow("img_rgb", img_rgb)
cv2.imshow("img_gray", img_gray)
cv2.imshow("img_hsv", img_hsv)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

Applying Blur

```python
img = cv2.imread("bird.jpg")
k_size = 7

img_blured = cv2.blur(img, (k_size, k_size))
img_gaussian_blured = cv2.GaussianBlur(img, (k_size, k_size), 5)
img_median_blured = cv2.medianBlur(img, k_size)

cv2.imshow("img", img)
cv2.imshow("img_blured", img_blured)
cv2.imshow("img_gaussian_blured", img_gaussian_blured)
cv2.imshow("img_median_blured", img_median_blured)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
